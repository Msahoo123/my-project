# **5. CNN Model Architecture**

## **5.1 Introduction to CNN Architecture**

Convolutional Neural Network (CNN) is one of the most important deep learning algorithms used in computer vision and image classification.

In the DeepFER project, CNN is used to identify and classify human emotions by analyzing facial images.

Unlike traditional machine learning algorithms, which often require manual feature extraction, CNN automatically learns important features from input images.

For example, when we provide a smiling face image to the CNN model, it learns different facial patterns, including the shape of the mouth, raised cheeks, and other visual features.

The model uses these learned features to predict the most likely emotion category.

Our CNN architecture is designed to classify seven emotions: Angry, Sad, Happy, Fear, Neutral, Disgust, and Surprise.

## **5.2 Main Layers of CNN Architecture**

### **1. Input Layer**

The input layer receives facial images from the dataset.

For our custom grayscale CNN model, the input image size is 48 × 48 pixels with one color channel.

Input Shape: (48, 48, 1)

The input image is passed to the convolutional layers for feature extraction.

### **2. Convolutional Layer**

The convolutional layer is responsible for extracting important facial features from the input image.

It uses filters or kernels to identify different patterns such as edges, curves, eyes, eyebrows, and mouth shapes.

For example:
- Early layers identify basic edges and lines.
- Middle layers identify facial parts.
- Deeper layers identify complex facial patterns.

We use different convolutional layers with 32, 64, and 128 filters to learn increasingly complex features.

### **3. ReLU Activation Function**

ReLU stands for Rectified Linear Unit.

It is an activation function used to introduce non-linearity into the neural network.

It converts negative values into zero while keeping positive values unchanged.

Mathematical formula:

ReLU(x) = max(0, x)

This helps the model learn complex patterns from facial images.

### **4. Max Pooling Layer**

Max Pooling reduces the spatial dimensions of feature maps while keeping the most important information.

For example, a 2 × 2 Max Pooling operation reduces the feature map size by approximately half in each spatial dimension.

Its main advantages are:
- Reduces computational complexity.
- Decreases the number of parameters in later layers.
- Helps reduce overfitting.
- Retains important image features.

### **5. Flatten Layer**

After feature extraction, the Flatten layer converts the two-dimensional feature maps into a one-dimensional vector.

This vector is passed to the fully connected layers for classification.

### **6. Fully Connected (Dense) Layer**

The Dense layer learns relationships between the extracted facial features.

It combines different patterns and helps the model identify which emotion category is most suitable.

We can use 256 neurons with ReLU activation in the first Dense layer.

### **7. Dropout Layer**

Dropout is a regularization technique used to reduce overfitting.

During training, it randomly disables some neurons so that the model does not depend too heavily on particular neurons.

For example, Dropout(0.5) randomly disables approximately 50% of the selected neurons during each training step.

### **8. Output Layer (Softmax)**

The final layer contains seven neurons because our project identifies seven emotion categories.

The Softmax activation function converts the model's output into probability values.

The emotion with the highest predicted probability is selected as the model's prediction.

The seven output categories are:

1. Angry
2. Sad
3. Happy
4. Fear
5. Neutral
6. Disgust
7. Surprise

Step 2. Import Libraries

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout,
    BatchNormalization
)

from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.metrics import classification_report
```

Step 3. Load Dataset

Arrange your dataset in this folder structure:

```python
DeepFER/
│
└── dataset/
    │
    ├── train/
    │   ├── angry/
    │   ├── sad/
    │   ├── happy/
    │   ├── fear/
    │   ├── neutral/
    │   ├── disgust/
    │   └── surprise/
    │
    ├── validation/
    │   ├── angry/
    │   ├── sad/
    │   ├── happy/
    │   ├── fear/
    │   ├── neutral/
    │   ├── disgust/
    │   └── surprise/
    │
    └── test/
        ├── angry/
        ├── sad/
        ├── happy/
        ├── fear/
        ├── neutral/
        ├── disgust/
        └── surprise/
```

Step 4. Data Preprocessing and Augmentation

```python
IMG_SIZE = (48, 48)
BATCH_SIZE = 64

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1,
    horizontal_flip=True
)

test_datagen = ImageDataGenerator(
    rescale=1./255
)

train_data = train_datagen.flow_from_directory(
    "dataset/train",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    batch_size=BATCH_SIZE,
    class_mode="categorical"
)

val_data = test_datagen.flow_from_directory(
    "dataset/validation",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    batch_size=BATCH_SIZE,
    class_mode="categorical"
)

test_data = test_datagen.flow_from_directory(
    "dataset/test",
    target_size=IMG_SIZE,
    color_mode="grayscale",
    batch_size=BATCH_SIZE,
    class_mode="categorical",
    shuffle=False
)
```

Step 5. Build CNN Model

```python
model = Sequential([

    Conv2D(32, (3, 3),
           activation='relu',
           input_shape=(48, 48, 1)),

    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(64, (3, 3), activation='relu'),

    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Conv2D(128, (3, 3), activation='relu'),

    BatchNormalization(),
    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(256, activation='relu'),

    Dropout(0.5),

    Dense(7, activation='softmax')
])

model.summary()
```

Step 6. Compile the Model

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

Step 7. Train the Model

```python
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=25
)
```

Step 8. Evaluate the Model

```python
test_loss, test_accuracy = model.evaluate(test_data)

print("Test Loss:", test_loss)

print("Test Accuracy:", test_accuracy)
```

Step 9. Classification Report

```python
predictions = model.predict(test_data)

y_pred = np.argmax(predictions, axis=1)

y_true = test_data.classes

class_names = list(test_data.class_indices.keys())

print(classification_report(
    y_true,
    y_pred,
    target_names=class_names
))
```

Step 10. Save the Trained Model

```python
model.save("deepfer_emotion_model.keras")

print("DeepFER Model Saved Successfully!")
```

Step 11. Display Training Accuracy Graph

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history['accuracy'],
    label='Training Accuracy'
)

plt.plot(
    history.history['val_accuracy'],
    label='Validation Accuracy'
)

plt.title("DeepFER Model Accuracy")

plt.xlabel("Epochs")

plt.ylabel("Accuracy")

plt.legend()

plt.show()
```

Step 12. Display Training Loss Graph

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history['loss'],
    label='Training Loss'
)

plt.plot(
    history.history['val_loss'],
    label='Validation Loss'
)

plt.title("DeepFER Model Loss")

plt.xlabel("Epochs")

plt.ylabel("Loss")

plt.legend()

plt.show()
```

### 2. Precision – 93%

Precision measures how many images predicted as a particular emotion actually belong to that emotion category.

For example, if the model predicts 100 images as Happy and 93 images are actually Happy, the precision is 93%.

Formula:

Precision = TP / (TP + FP)

Where:
- TP = True Positive
- FP = False Positive

In our project, precision helps us understand how reliably the model predicts different facial emotion categories.

### 3. Recall – 93%

Recall measures how many actual facial images belonging to a particular emotion category are correctly identified by the model.

For example, if the dataset contains 100 actual Happy images and the model correctly identifies 93 of them, the recall is 93%.

Formula:

Recall = TP / (TP + FN)

Where:
- TP = True Positive
- FN = False Negative

A higher recall means the model can identify more of the actual emotion samples.

### 4. F1-Score – 93%

F1-Score combines Precision and Recall into a single evaluation metric.

It helps us understand the balance between correctly identifying emotions and avoiding incorrect predictions.

Formula:

F1-Score = 2 × (Precision × Recall) / (Precision + Recall)

The sample F1-score is 93%, indicating a good balance between precision and recall.

Overall, these four evaluation metrics help us understand the classification performance of our DeepFER model.

##**10.9 Real-Time Demo – Python Implementation**
The following code demonstrates how to connect your trained DeepFER model to a webcam.

Python Code:

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# Load trained model
model = load_model("DeepFER_Final_Model.keras")

# Emotion categories
emotion_labels = [
    "Angry",
    "Disgust",
    "Fear",
    "Happy",
    "Sad",
    "Surprise",
    "Neutral"
]

# Load face detector
face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

# Start webcam
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    faces = face_detector.detectMultiScale(
        gray,
        scaleFactor=1.3,
        minNeighbors=5
    )

    for (x, y, w, h) in faces:

        face = gray[y:y+h, x:x+w]

        face = cv2.resize(face, (48, 48))

        face = face.astype("float32") / 255.0

        face = np.expand_dims(face, axis=-1)

        face = np.expand_dims(face, axis=0)

        # Predict emotion
        prediction = model.predict(
            face,
            verbose=0
        )

        emotion_index = np.argmax(prediction)

        emotion = emotion_labels[emotion_index]

        confidence = prediction[0][emotion_index]

        # Draw face rectangle
        cv2.rectangle(
            frame,
            (x, y),
            (x+w, y+h),
            (0, 255, 0),
            2
        )

        # Display emotion
        text = f"{emotion}: {confidence:.2f}"

        cv2.putText(
            frame,
            text,
            (x, y-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 0),
            2
        )

    cv2.imshow(
        "DeepFER - Real-Time Emotion Recognition",
        frame
    )

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()

cv2.destroyAllWindows()
```

**Important:** This example assumes your trained model accepts 48 × 48 grayscale images and uses the emotion-label order shown in the code. Change the input preprocessing and label order to match your actual training configuration.
